```python
from transformers import VitsModel, AutoTokenizer
import torch
import scipy.io.wavfile as wavfile
import soundfile as sf
import numpy as np

model_name = "facebook/mms-tts-hin"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = VitsModel.from_pretrained(model_name)

with open("bhojpuri_translation.txt", "r", encoding="utf-8") as f:
    full_text = f.read()

chunk_size = 300
chunks = [full_text[i:i+chunk_size] for i in range(0, len(full_text), chunk_size)]

all_audio = []

for idx, chunk in enumerate(chunks):
    print(f"Processing chunk {idx+1}/{len(chunks)}")

    inputs = tokenizer(chunk, return_tensors="pt")

    with torch.no_grad():
        output = model(**inputs).waveform

    audio_chunk = output.squeeze().cpu().numpy()
    all_audio.append(audio_chunk)

final_audio = np.concatenate(all_audio)

sf.write("output_LRL_cloned.wav", final_audio, 22050)

print("Full MMS synthesis complete!")
```
